## Q1(a) Reverse a NumPy array

````python
arr = np.array([1, 2, 3, 6, 4, 5])

reversed_arr = arr[::-1]

print("Original array :", arr)
print("Reversed array :", reversed_arr)
````

### Answer
The reversed array is **[5, 4, 6, 3, 2, 1]**.

`[::-1]` uses NumPy/Python slicing with a step of `-1`, so the elements are read from the last position to the first.


## Q1(b) Flatten an array using two NumPy methods

````python
array1 = np.array([[1, 2, 3],
                   [2, 4, 5],
                   [1, 2, 3]])

# Method 1: flatten()
flatten_method_1 = array1.flatten()

# Method 2: ravel()
flatten_method_2 = array1.ravel()

print("Original array:")
print(array1)
print("\nUsing flatten():", flatten_method_1)
print("Using ravel()  :", flatten_method_2)
````

### Answer
Both methods produce:

`[1, 2, 3, 2, 4, 5, 1, 2, 3]`

- `flatten()` returns a flattened copy.
- `ravel()` returns a flattened array and may return a view when possible.

## Q1(d) Most frequent value and its indices

````python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

def most_frequent_values_with_indices(array):
    values, counts = np.unique(array, return_counts=True)
    max_count = counts.max()
    most_frequent = values[counts == max_count]

    return [
        (int(value), int(max_count), np.where(array == value)[0])
        for value in most_frequent
    ]

print("For x:")
for value, count, indices in most_frequent_values_with_indices(x):
    print(f"Value = {value}, Frequency = {count}, Indices = {indices}")

print("\nFor y:")
for value, count, indices in most_frequent_values_with_indices(y):
    print(f"Value = {value}, Frequency = {count}, Indices = {indices}")
````

### Answer
- For `x`, the most frequent value is **1**, appearing **5 times** at indices **[0, 5, 7, 8, 9]**.
- For `y`, there is a tie:
  - Value **1** appears 3 times at **[0, 1, 2]**.
  - Value **3** appears 3 times at **[4, 8, 9]**.

### Answer

3. **Eigenvectors:** the columns of the matrix printed above correspond to those eigenvalues.
4. **Inverse:** approximately

```text
[[ 0.024041 -0.009112 -0.004447]
 [-0.016679  0.029669  0.002479]
 [ 0.006312 -0.016037  0.012174]]
```

5. **Determinant:** `137180`

## Q1(g) Matrix multiplication and covariance

### (i) 2 × 2 matrices

````python
p = np.array([[1, 2],
              [2, 3]])

q = np.array([[4, 5],
              [6, 7]])

matrix_product_1 = p @ q
covariance_1 = np.cov(p.flatten(), q.flatten())

print("Matrix multiplication (p @ q):")
print(matrix_product_1)

print("\nCovariance matrix:")
print(covariance_1)
````

### (ii) 3 × 2 and 2 × 3 matrices

````python
p = np.array([[1, 2],
              [2, 3],
              [4, 5]])

q = np.array([[4, 5, 1],
              [6, 7, 2]])

matrix_product_2 = p @ q
covariance_2 = np.cov(p.flatten(), q.flatten())

print("Matrix multiplication (p @ q):")
print(matrix_product_2)

print("\nCovariance matrix:")
print(covariance_2)
````

### Answer
For both parts, matrix multiplication is performed with `@`.

For covariance, the matrices are flattened because covariance is computed between two one-dimensional sets of observations. `np.cov()` returns a 2 × 2 covariance matrix:
- diagonal values are the variances,
- off-diagonal values are the covariance between the two inputs.


## Q1(h) Inner, outer and Cartesian product

````python
x = np.array([[2, 3, 4],
              [3, 2, 9]])

y = np.array([[1, 5, 0],
              [5, 10, 3]])

inner_product = np.inner(x, y)
outer_product = np.outer(x, y)

# Cartesian product of all elements after flattening
cartesian_product = np.array(list(product(x.flatten(), y.flatten())))

print("Inner product:")
print(inner_product)

print("\nOuter product:")
print(outer_product)

print("\nCartesian product:")
print(cartesian_product)
````

### Answer
- `np.inner(x, y)` gives a **2 × 2** matrix of row-wise inner products.
- `np.outer(x, y)` first treats both inputs as flattened arrays and produces a **6 × 6** outer-product matrix.
- The Cartesian product contains **36 pairs** because each of the 6 elements of `x` is paired with each of the 6 elements of `y`.


# Q2 — NumPy Mathematics and Statistics

## Q2(a) Absolute value, percentiles, mean, median and standard deviation

````python
array = np.array([
    [1, -2, 3],
    [-4, 5, -6]
])

print("Element-wise absolute value:")
print(np.abs(array))

print("\n25th, 50th, 75th percentiles — flattened:")
print(np.percentile(array.flatten(), [25, 50, 75]))

print("\n25th, 50th, 75th percentiles — each column:")
print(np.percentile(array, [25, 50, 75], axis=0))

print("\n25th, 50th, 75th percentiles — each row:")
print(np.percentile(array, [25, 50, 75], axis=1))

print("\nMean — flattened:", np.mean(array))
print("Median — flattened:", np.median(array))
print("Std Dev — flattened:", np.std(array))

print("\nMean — each column:", np.mean(array, axis=0))
print("Median — each column:", np.median(array, axis=0))
print("Std Dev — each column:", np.std(array, axis=0))

print("\nMean — each row:", np.mean(array, axis=1))
print("Median — each row:", np.median(array, axis=1))
print("Std Dev — each row:", np.std(array, axis=1))
````

### Answer / important note
NumPy's default standard deviation is the **population standard deviation** (`ddof=0`), which is what is reported above.

For the column statistics:
- Mean = `[-1.5, 1.5, -1.5]`
- Median = `[-1.5, 1.5, -1.5]`
- Standard deviation = `[2.5, 3.5, 4.5]`

For the row statistics:
- Mean = `[0.0, -1.666667]`
- Median = `[-1.0, -4.0]`
- Standard deviation = `[2.160247, 4.784233]`


## Q2(b) Floor, ceiling, truncated and rounded values

````python
a = np.array([-1.8, -1.6, -0.5, 0.5, 1.6, 1.8, 3.0])

print("Original :", a)
print("Floor    :", np.floor(a))
print("Ceiling  :", np.ceil(a))
print("Truncated:", np.trunc(a))
print("Rounded  :", np.round(a))
````

### Answer
- **Floor:** `[-2, -2, -1, 0, 1, 1, 3]`
- **Ceiling:** `[-1, -1, 0, 1, 2, 2, 3]`
- **Truncated:** `[-1, -1, 0, 0, 1, 1, 3]`
- **Rounded:** `[-2, -2, 0, 0, 2, 2, 3]`

`np.round()` follows NumPy's rounding behavior; for halfway cases, NumPy uses rounding to the nearest even value.

# Q4 — Image to NumPy Array and Text File

The assignment asks for a function `img_to_array(path)` that reads an image and saves its array representation as a text file, with separate handling for RGB and grayscale images, followed by loading the saved file into a Jupyter notebook.

The implementation below:
- detects whether the image is RGB or grayscale,
- converts it to a NumPy array,
- stores the shape and mode in the text-file header,
- reshapes RGB data into 2D before `np.savetxt()` because `np.savetxt()` expects a 1D/2D array,
- reconstructs the original shape when the text file is loaded.


````python
from PIL import Image

def img_to_array(path):
    """Read an image, convert it to a NumPy array, and save it as text.

    Returns
    -------
    output_path : str
        Path of the generated text file.
    array : np.ndarray
        Image represented as a NumPy array.
    mode : str
        'RGB' or 'GRAY'.
    """
    path = Path(path)

    with Image.open(path) as image:
        if image.mode == "RGB":
            mode = "RGB"
            image_array = np.array(image)
        else:
            # Treat all non-RGB images as grayscale for this assignment.
            mode = "GRAY"
            image_array = np.array(image.convert("L"))

    output_path = path.with_name(path.stem + "_array.txt")

    # np.savetxt supports 1D/2D arrays, so RGB data is flattened
    # into (height, width * 3) while its original shape is saved in the header.
    if image_array.ndim == 3:
        h, w, c = image_array.shape
        data_to_save = image_array.reshape(h, w * c)
        shape_text = f"{h},{w},{c}"
    else:
        h, w = image_array.shape
        data_to_save = image_array
        shape_text = f"{h},{w}"

    np.savetxt(
        output_path,
        data_to_save,
        fmt="%d",
        header=f"mode={mode};shape={shape_text}",
        comments="# "
    )

    return str(output_path), image_array, mode


# Example:
# output_path, image_array, mode = img_to_array("sample.jpg")
# print("Mode:", mode)
# print("Shape:", image_array.shape)
# print("Saved to:", output_path)
````

## Q4(b) Load the saved text file into Jupyter Notebook

````python
def load_image_array_from_text(txt_path):
    """Load an image array saved by img_to_array()."""
    txt_path = Path(txt_path)

    # Read the metadata from the first line.
    with open(txt_path, "r") as file:
        header = file.readline().strip()

    # Example header:
    # # mode=RGB;shape=100,200,3
    header = header.lstrip("# ").strip()
    mode_part, shape_part = header.split(";")
    mode = mode_part.split("=", 1)[1]
    shape = tuple(map(int, shape_part.split("=", 1)[1].split(",")))

    data = np.loadtxt(txt_path, dtype=np.uint8)

    if mode == "RGB":
        data = data.reshape(shape)
    else:
        data = data.reshape(shape)

    return data, mode


# Example:
# loaded_array, mode = load_image_array_from_text("sample_array.txt")
# print("Loaded mode :", mode)
# print("Loaded shape:", loaded_array.shape)
# print(loaded_array)
````

### Q4 explanation

For an RGB image, the NumPy shape is generally:

`(height, width, 3)`

where the last dimension stores the Red, Green and Blue channel values.

For a grayscale image, the shape is generally:

`(height, width)`

The text file contains metadata so that the saved numerical data can be reconstructed into its original image-array shape.

### Example workflow

```python
output_path, image_array, mode = img_to_array("my_image.jpg")

loaded_array, mode = load_image_array_from_text(output_path)

print(mode)
print(loaded_array.shape)
```
