```python
# Enter your details before running the remaining cells.

ROLL_NUMBER = input("Enter your roll number: ").strip()

# These fields can be edited directly for your submission.
NAME = "Pushkar Sheokand"
BRANCH = "COPC"
AGE = 20
CITY = "Karnal"
CGPA = 8.6

# Basic validation
if not ROLL_NUMBER.isdigit():
    raise ValueError("Roll number must contain digits only.")

if len(ROLL_NUMBER) < 8:
    raise ValueError("The assignment requires at least 8 digits in the roll number.")

digits = [int(d) for d in ROLL_NUMBER]
first_8_digits = digits[:8]

print("Roll number:", ROLL_NUMBER)
print("First 8 digits:", first_8_digits)
```

# Q1 — Lists

The assignment asks us to extract the individual digits of the roll number, multiply each digit by 10, and create a list `L`. It then covers `append()`, `insert()`, `remove()`, `pop()`, sorting, slicing and list comprehension.


## Q1(i) Create and print L

```python
# Multiply every digit of the roll number by 10.
L = [digit * 10 for digit in digits]

print("L =", L)
```

## Q1(ii) Add two numbers using append() and insert()

```python
# append() adds the chosen value at the end of the list.
L.append(75)
print("After append(75):", L)

# insert(position, value) places the chosen value at a specific index.
L.insert(2, 25)
print("After insert(2, 25):", L)
```

## Q1(iii) Remove two elements using remove() and pop()

```python
# remove(value) removes the first occurrence of the specified value.
# We choose the value 75, which was added above.
if 75 in L:
    L.remove(75)
print("After remove(75):", L)

# pop(index) removes and returns the element at the specified index.
removed_value = L.pop(2)
print(f"After pop(2): {L}")
print("Value removed by pop():", removed_value)
```

## Q1(iv) Sort ascending and descending

```python
L.sort()
print("Ascending order :", L)

L.sort(reverse=True)
print("Descending order:", L)
```

## Q1(v) Slicing — first three and last three elements

```python
print("First three elements:", L[:3])
print("Last three elements :", L[-3:])
```

## Q1(vi) List comprehension — elements greater than average

```python
average_L = sum(L) / len(L)

greater_than_average = [value for value in L if value > average_L]

print("Average of L:", average_L)
print("Elements greater than average:", greater_than_average)
```

# Q2 — Tuples

The first 8 values from `L` are placed into a tuple called `scores`. The questions demonstrate tuple indexing, counting, reversing, searching, immutability and unpacking with `*`.


## Q2 — Create the tuple

```python
scores = tuple(L[:8])

print("scores =", scores)
```

## Q2(i) Highest/lowest score and their information

```python
highest_score = max(scores)
highest_index = scores.index(highest_score)

lowest_score = min(scores)
lowest_count = scores.count(lowest_score)

print("Highest score:", highest_score)
print("Index of highest score:", highest_index)
print("Lowest score:", lowest_score)
print("Number of times lowest score appears:", lowest_count)
```

## Q2(ii) Reverse the tuple and return it as a list

```python
# Tuples are immutable, so they cannot be reversed in place.
# We create a reversed iterator and convert it into a new list.
reversed_scores = list(reversed(scores))

print("Reversed tuple as a list:", reversed_scores)
```

## Q2(iii) Search for a user-entered score

```python
user_score = int(input("Enter a score to search for: "))

if user_score in scores:
    print("First occurrence index:", scores.index(user_score))
else:
    print("Score is not present in the tuple.")
```

## Q2(iv) Attempt to modify a tuple

The assignment specifically asks us to run an invalid tuple assignment and capture the Python error.


```python
try:
    # This is intentionally invalid and is included to demonstrate
    # the immutability of tuples.
    scores[0] = 100
except TypeError as error:
    print("Captured error:", error)
    print("Explanation: tuples are immutable, so their existing elements cannot be changed; lists allow item assignment.")
```

## Q2(v) Tuple unpacking using `*`

```python
first_score, second_score, *remaining_scores = scores

print("First score :", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

# Q3 — Random Numbers

The assignment requires `random.seed(your_roll_number)` before generating numbers. This makes the random list reproducible for the same student while producing different results for a different seed.


## Q3(i) Generate 100 random numbers between 100 and 900

```python
import random
from collections import Counter

# The roll number is used as the random seed exactly as required.
random.seed(int(ROLL_NUMBER))

random_numbers = [random.randint(100, 900) for _ in range(100)]

print("Generated 100 random numbers:")
print(random_numbers)
```

## Q3(ii) Count and print all odd numbers

```python
odd_numbers = [number for number in random_numbers if number % 2 != 0]

print("Odd numbers:")
print(odd_numbers)
print("Count of odd numbers:", len(odd_numbers))
```

## Q3(iii) Count and print all even numbers

```python
even_numbers = [number for number in random_numbers if number % 2 == 0]

print("Even numbers:")
print(even_numbers)
print("Count of even numbers:", len(even_numbers))
```

## Q3(iv) Count and print all prime numbers

```python
def is_prime(number):
    if number < 2:
        return False

    # Test divisors only up to sqrt(number).
    divisor = 2
    while divisor * divisor <= number:
        if number % divisor == 0:
            return False
        divisor += 1

    return True


prime_numbers = [number for number in random_numbers if is_prime(number)]

print("Prime numbers:")
print(prime_numbers)
print("Count of prime numbers:", len(prime_numbers))
```

## Q3(v) Most frequently occurring number

```python
frequency = Counter(random_numbers)
most_frequent_number, highest_frequency = frequency.most_common(1)[0]

print("Most frequent number:", most_frequent_number)
print("Number of occurrences:", highest_frequency)
```

# Q4 — Sets

The assignment uses the first 8 digits of the roll number before multiplying them by 10. Two sets are created:

- `A = {digit × 7}`
- `B = {digit × 9}`

Because sets store only unique values, repeated roll-number digits may produce fewer than 8 elements.


## Q4 — Create sets A and B

```python
A = {digit * 7 for digit in first_8_digits}
B = {digit * 9 for digit in first_8_digits}

print("Set A =", A)
print("Set B =", B)
```

## Q4(vi) Union

```python
union_AB = A.union(B)

print("A union B =", union_AB)
```

## Q4(vii) Intersection

```python
intersection_AB = A.intersection(B)

print("A intersection B =", intersection_AB)
```

## Q4(viii) A − B and B − A using difference()

```python
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B =", A_minus_B)
print("B - A =", B_minus_A)

# difference() gives the elements present in one set but not the other,
# while symmetric_difference() gives the elements present in either set but not both.
```

## Q4(ix) Symmetric difference

```python
symmetric_difference_AB = A.symmetric_difference(B)

print("Symmetric difference of A and B =", symmetric_difference_AB)
```

## Q4(x) Subset and superset checks

```python
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

## Q4(xi) Remove a user-entered value using discard()

```python
X = int(input("Enter a value to discard from set A: "))

A.discard(X)

print("Set A after discard:", A)

# discard() is safer than remove() because discard() does not raise
# a KeyError when the requested value is not present in the set.
```

# Q5 — Dictionaries

The assignment requires a dictionary containing personal/student information and then demonstrates renaming keys, adding and updating entries, deletion using `pop()` and `del`, iteration with `.items()`, safe key checking, dictionary merging and dictionary comprehension.


## Q5 — Create the original dictionary

```python
my_dict = {
    "name": NAME,
    "roll_no": ROLL_NUMBER,
    "branch": BRANCH,
    "age": AGE,
    "city": CITY
}

print("Original dictionary:")
print(my_dict)
```

## Q5(i) Rename `city` to `location`

```python
# pop() removes the old key and returns its value.
my_dict["location"] = my_dict.pop("city")

print("After renaming city to location:")
print(my_dict)
```

## Q5(ii) Add the `cgpa` key

```python
my_dict["cgpa"] = CGPA

print("After adding CGPA:")
print(my_dict)
```

## Q5(iii) Increase age by 1

```python
my_dict["age"] += 1

print("After increasing age by 1:")
print(my_dict)
```

## Q5(iv) Delete `branch` using pop() and del

```python
# Make two independent copies so both deletion methods
# can be demonstrated without affecting each other.
dict_pop = my_dict.copy()
dict_del = my_dict.copy()

removed_branch = dict_pop.pop("branch")

del dict_del["branch"]

print("Dictionary after pop():")
print(dict_pop)
print("Value returned by pop():", removed_branch)

print("\nDictionary after del:")
print(dict_del)

# pop() removes the key and returns its value, while del removes the
# key-value pair but does not return the removed value.
```

## Q5(v) Iterate using `.items()`

```python
for key, value in my_dict.items():
    print(f"{key} → {value}")
```

## Q5(vi) Safely check whether `email` exists

```python
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key is not present in the dictionary.")
```

## Q5(vii) Merge with a friend's dictionary

```python
friend_dict = {
    "name": "Fictional Friend",
    "roll_no": "FRIEND001",
    "branch": "CSE",
    "age": 21,
    "city": "Delhi"
}

# The original dictionary has been updated to contain "location".
# The assignment asks for merging using {**dict1, **dict2}.
merged_dict = {**my_dict, **friend_dict}

print("Friend dictionary:")
print(friend_dict)

print("\nMerged dictionary:")
print(merged_dict)

# When both dictionaries contain the same key, values from the dictionary
# appearing later in the unpacking expression win.
```

## Q5(viii) Dictionary comprehension — only string values

```python
string_values_dict = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("Key-value pairs whose values are strings:")
print(string_values_dict)
```
