```python
import pandas as pd
ROLL_NO="1024170381"

fixed_entries = [
    {"question": "what is the annual fee",
     "answer": "The annual fee is Rs 500.",
     "keywords": "fee cost price charge",
     "category": "billing"},
    {
      "question": "how to reset password",
      "answer": "Go to Settings > Reset Password.",
      "keywords": "password reset login",
      "category": "account"
    },
    {
      "question": "what are your working hours",
      "answer": "We are open 9 AM to 5 PM.",
      "keywords": "hours timing open time",
      "category": "general"
    },
    {
      "question": "how can i pay the fee",
      "answer": "You can pay via UPI, card, or net banking.",
      "keywords": "pay payment upi fee",
      "category": "billing"
    }
]

for i in range(2):
  d = int(ROLL_NO[i-2])
  category=["billing", "account", "general"][d%3]
  match category:
    case "billing":
      fixed_entries.append({
        "question": "is there a refund policy",
        "answer": "Yes, you can request a full refund within 7 days of purchase.",
        "keywords": "refund money back return cancel",
        "category": "billing"
      })
    case "account":
      fixed_entries.append({
        "question": "how to change my email address",
        "answer": "Update your email under Account Settings > Security.",
        "keywords": "email change update profile personal",
        "category": "account"
      })
    case "general":
      fixed_entries.append({
        "question": "where is your office located",
        "answer": "Our headquarters is at 101 Innovation Way, Tech District.",
        "keywords": "location office address building map headquarter",
        "category": "general"
      })

df = pd.DataFrame(fixed_entries)
print(df)
```

```python
query = input("Enter query: ")
def score_query(q, df):
  query_words = set(q.lower().split())
  results = []
  for index, row in df.iterrows():
    keywords = set(row["keywords"].lower().split())
    score = len(query_words & keywords)
    if score > 0:
      results.append((score, index, row["question"]))

  results.sort(reverse=True)
  return results

res = score_query(query, df)
for s, i, q in res:
  print("Score: ", s)
  print("Question: ", q)
```

```python
def same_category(category_name, df):
  filtered = df[df["category"] == category_name]
  return filtered["question"].tolist()

print(same_category("general", df))



```

```python
row = int(input("Enter row: "))
keyword = input("Enter a keyword: ")
col = df.columns.get_loc("keywords")
current_keywords = df.iloc[row, col]
df.iloc[row, col] = f"{current_keywords} {keyword}"
csv_file_name = f"{ROLL_NO}_faq_data.csv"
df.to_csv(csv_file_name, index=False)
```
